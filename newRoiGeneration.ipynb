```python
import pandas as pd

# -----------------------------
# Step 1: Read ROI file
# -----------------------------
roi = pd.read_excel('rois.xlsx', header=None)
roi_list = roi[0].astype(str).tolist()


# -----------------------------
# Step 2: Read sampling files
# -----------------------------
s1 = pd.read_excel('All_tissue_specific_dysregulated_p10_no_exchange_annotated.xlsx')
s2 = pd.read_excel('All_tissue_specific_dysregulated_p15_no_exchange_annotated.xlsx')


# -----------------------------
# Step 3: Convert sampling → tissue_rxn format
# -----------------------------
s1_combined = (s1['tissue'].astype(str) + "_" + s1['rxns'].astype(str)).tolist()
s2_combined = (s2['tissue'].astype(str) + "_" + s2['rxns'].astype(str)).tolist()


# -----------------------------
# Step 4: Combine all
# -----------------------------
all_rois = roi_list + s1_combined + s2_combined


# -----------------------------
# Step 5: Remove transport reactions
# -----------------------------
filtered_rois = [
    r for r in all_rois
    if ("Tr" not in r and "[" not in r and "]" not in r)
]


# -----------------------------
# Step 6: Remove duplicates
# -----------------------------
final_rois = list(set(filtered_rois))


# -----------------------------
# Step 7: Save final file
# -----------------------------
pd.Series(final_rois).to_csv('final_ROIs.txt', index=False, header=False)

print("✅ Done: Transport reactions removed + final_ROIs.txt created")
```
